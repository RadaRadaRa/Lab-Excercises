```python
import requests
import json
```

```python
#This rather simple and silly api accesses a database full of insults.
#It does not require payment nor does it require a key to use. 
```

```python
#The base url is "https://evilinsult.com"
```

```python
#https://en.wikipedia.org/wiki/List_of_ISO_639_language_codes
#Website to get language 2 letter iso code.
```

```python
#This api call pulls a random insult from their database
#It has two optional parameters, the language of the insult, and the type of the response. 
#The language of the insult defaults to english if not specified. It takes 2 letter language iso codes. 
#The type of the response defaults to plain text if not specified. It supports text, xml, and json.
```

```python
#In this example, we pull a random insult in english in a json format from the database. 
englishinsult=requests.get(f"https://evilinsult.com/generate_insult.php?lang=en&type=json")
englishinsultjson=englishinsult.json()
englishinsultjson

#The json information contains a variety of information about the insult.
#'number' signifies the id number of the insult.
#'language' contains the language of the insult.
#'insult' contains the insult itself.
#'created' tells the user when the api call was made.
#'shown' tells the user how many times this insult has been called.
#'createdby' tells the user who created the insult.
#'active' appears to signify whether this insult is avaliable to be called.
#'comment' will return any comments other user have about this particular insult
```

```python
#In this example, we pull a random insult in spanish in text from the database. 
spanishinsult=requests.get(f"https://evilinsult.com/generate_insult.php?lang=es&type=text")
spanishinsulttext=spanishinsult.text
spanishinsulttext
#To get the data from the api call for insults in text format, we need to convert the api call to text using text().
```
