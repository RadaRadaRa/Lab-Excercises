```python
import requests
import json
import csv
import pandas as pd
```

```python
publicholidays=requests.get(f"https://date.nager.at/api/v3/AvailableCountries")
publicholidaysjson=publicholidays.json()
publicholidaysjson
```

```python
csv_file = open("Holidays.csv", "a", newline="", encoding="utf-8")
csv_writer = csv.writer(csv_file)
csv_writer.writerow(["Country","Total Number of Holidays"])
csv_file.close()
year="2024"
for i in publicholidaysjson:
    csv_file = open("Holidays.csv", "a", newline="", encoding="utf-8")
    csv_writer = csv.writer(csv_file)
    countryCode=i["countryCode"]
    publicholidays=requests.get(f"https://date.nager.at/api/v3/PublicHolidays/{year}/{countryCode}")
    publicholidaysjson=publicholidays.json()
    csv_writer.writerow([i["name"],len(publicholidaysjson)])
    csv_file.close()
sheet=pd.read_csv("Holidays.csv")
```
