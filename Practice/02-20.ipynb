```python
data_file = open("census_data.csv", "w", newline="", encoding="utf-8")
csv_writer = csv.writer(data_file)
csv_writer.writerow(["Name", "Pop", "HISP", "State"])


for list in json_filejson:
    if list == json_filejson[0]:
        pass
    else:
        name = list[0]
        pop = list[1]
        hisp = list[2]
        state = list[3]
        
        csv_writer.writerow([name, pop, hisp, state])


data_file.close()
```
