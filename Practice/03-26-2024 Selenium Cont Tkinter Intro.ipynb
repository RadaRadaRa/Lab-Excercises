```python
from bs4 import BeautifulSoup
import html5lib
import requests
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
source = requests.get("https://www.scrapethissite.com/pages/").text
soup = BeautifulSoup(source, "lxml")

for element in soup.find_all("div", class_="page"):
  desc = element.p.text.strip()
  title = element.h3.text.strip()

  print(title, desc)
```

```python
driver=webdriver.Chrome()
driver.get("https://www.scrapethissite.com")
WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.ID, "nav-sandbox"))
)
sandbox=driver.find_element(By.ID,"nav-sandbox")
sandbox.click()
WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.CSS_SELECTOR, "div.page"))
)
for i in driver.find_elements(By.CSS_SELECTOR, "div.page"):
    print(i.find_element(By.TAG_NAME,"h3").text,"-",i.find_element(By.TAG_NAME,"p").text)  
driver.close()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
import time

driver = webdriver.Chrome()

driver.get("https://data.gov/")

search = driver.find_element(By.ID, "search-field")
search.send_keys("Climate Change")
search.send_keys(Keys.RETURN)

try:
    element = WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.ID, "field-order-by"))
    )  

    menu = driver.find_element(By.ID, "field-order-by")

    actions = ActionChains(driver)
    actions.move_to_element(menu)
    actions.click(menu)
    
    menu_item = driver.find_element(By.XPATH, "/html/body/div[2]/div/div[1]/div/form/div[1]/select/option[2]")
    menu_item.click()
    
    time.sleep(5)

finally:
    driver.quit()
```
