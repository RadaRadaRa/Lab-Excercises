```python
def pod(date=""):
    API_key="ObsbJXcSdM7ckQAll9qqccaG2XblPXGIj5tRxBRV"
    pod=requests.get(f"https://api.nasa.gov/planetary/apod?api_key={API_key}&date={date}")
    podjson=pod.json()
    urllib.request.urlretrieve(podjson['hdurl'], "gfg.png")
    img=Image.open("gfg.png")
    print(podjson['hdurl'])
    return img
```
