```python
import requests
import json
root=Tk()
root.geometry("1080x633")

# lat = "42.098701"
# lon = "-75.912537"
label_lat = Label(root, text="Lat").pack()
lat=StringVar()
lat.set("")
lat_entry = Entry(root, width=35, borderwidth=5, textvariable=lat)
lat_entry.pack()
label_lon = Label(root, text="Lon").pack()
lon=StringVar()
lon.set("")
lon_entry = Entry(root, width=35, borderwidth=5,textvariable=lon)
lon_entry.pack()

def weekly_forecast(lat, lon):
    weather = requests.get(f"https://api.weather.gov/points/{lat},{lon}")

    json = weather.json()
    forecast = json["properties"]["forecast"]
    daily_forecast = requests.get(forecast).json()

    for section in daily_forecast["properties"]["periods"]:
        day = section["name"]
        temp = section["temperature"]
        detail = section["detailedForecast"]
        Lable_maker=Label(root, text=f"{day}\n{temp}\n{detail}")
        Lable_maker.pack()
    
button=Button(root, text="start",command=lambda:weekly_forecast(lat.get(),lon.get())) #use lambda to run functions with parameters
button.pack()
destroy=Button(root,text="SELF DESTRUCT",command=root.destroy)
destroy.pack()
root.mainloop() 
```
