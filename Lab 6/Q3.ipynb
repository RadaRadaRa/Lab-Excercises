```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
#specifies chrome to be my webdriver
driver=webdriver.Chrome()
#grabs food52 website 
driver.get("https://food52.com/recipes")
#waits for driver to locate search bar
WebDriverWait(driver, 1).until(
    EC.presence_of_element_located((By.ID,"search-panelInput"))
)
#finds search bar
search=driver.find_element(By.ID,"search-panelInput")
#clears the search bar
search.clear()
#inputs onion into the sarch bar and presses enter
search.send_keys("onion"+Keys.ENTER)
#sets counter
counter=0
#while loop that grabs the first 5 recipe names
while counter<5:
    WebDriverWait(driver, 2).until(EC.presence_of_element_located((By.CLASS_NAME,"collectable__name")))
    print(driver.find_elements(By.CLASS_NAME,"collectable__name")[counter].text)
    counter +=1
```
