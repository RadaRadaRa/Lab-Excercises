```python
#Populates file with a row w/ data
class Rorschach_csv:
    def __init__(self,name):
        self.name=name
        csv_file = open(self.name+"_Rorschach_data.csv", "w", newline="", encoding="utf-8")
        csv_writer = csv.writer(csv_file)
        csv_writer.writerow(["Image", "Data"])
        csv_file.close()
    def rorschach_csv_filler(self,number_of_desired_entries):
        self.number_of_desired_entries=number_of_desired_entries
        for i in range(self.number_of_desired_entries):
            csv_file = open(self.name+"_Rorschach_data.csv", "a", newline="", encoding="utf-8")
            csv_writer = csv.writer(csv_file)
            image_name=choose_image()
            user_input=input("Briefly describe what you see in the image: ")
            csv_writer.writerow([image_name, user_input])
            csv_file.close()
            sheet=pd.read_csv(self.name+"_Rorschach_data.csv")
        return sheet
```
