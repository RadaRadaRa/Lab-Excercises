```python
import requests
import json
from tkinter import *
from PIL import ImageTk, Image
```

```python
root = Tk()
root.geometry("800x900") 
root.title("Spoontacular Search!")

key="198eadf675b34df284028f41d929cbe2"

def Results(ingredient):
    food=requests.get(f"https://api.spoonacular.com/recipes/findByIngredients?ingredients={ingredient}&apiKey={key}")
    food_json=food.json()
    Text_window.delete("1.0", "end")
    for i in food_json:
        id_of_recipe=i["id"]  
        json=requests.get(f"https://api.spoonacular.com/recipes/{id_of_recipe}/information?apiKey={key}").json()
        title=json["title"]
        price=json["pricePerServing"]
        Text_window.insert(END, f"{title} - ${round(price / 100, 2)}\n ")

def chosen_option():
    ingredient=ingredient_select.get()
    Results(ingredient)

left=Frame(root)
left.grid(row=0,column=0)
right=Frame(root)
right.grid(row=0,column=1)
Text_window=Text(root,height=70)
Text_window.grid(columnspan=2)

ingredient_select=StringVar()
ingredient_select.set("Select an Ingredient")

ingredients_list=["onion","potato","egg","tomato","garlic"]
common_label=Label(right,text="Common Household Ingredients").pack()
commondown=OptionMenu(right, ingredient_select,*ingredients_list).pack()

choose_button = Button(right, text="Select", command=chosen_option)
choose_button.pack()

ingredient=StringVar()
ingredient.set("")
label_ingre=Label(left, text="Search recipes by ingredients").pack()
entry_ingre=Entry(left, width=35, borderwidth=5, textvariable=ingredient).pack()
button=Button(left, font=12, text="Get recipe", cursor="star",command= lambda: Results(ingredient.get()))
button.pack()


root.mainloop()
```
