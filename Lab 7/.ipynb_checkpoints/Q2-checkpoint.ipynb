I thought it would be fun to add a little advice column that would advise the user what to do depending on the average temperature over the week. Using if else statements, it determines the advice and picture to display. Then I had to import ImageTk and Image from Python Image Library to display the images because using the built in image parameter in tkinter was really giving me trouble. I used a try except to remove the labels so that the program can be reusable. More details about my changes can be found in the comments in the code. 

```python
import requests
import json
from tkinter import *
from PIL import ImageTk, Image

root = Tk()
root.geometry("800x900") 
root.title("Weather App")



lat = StringVar() # lat = "42.098701"
lat.set("")
lon = StringVar() # lon = "-75.912537"
lon.set("")
city_select = StringVar()
city_select.set("Select a City")

#source https://forecast.weather.gov/stations.php?foo=0
city_list = ["Binghamton", "New York", "Chicago", "Atlanta", "Denver"]

#list syntax: [lat, lon]
city_latlon = {"Binghamton": ["42.09", "-75.91"],
               "New York": ["40.78", "-73.96"],
              "Chicago": ["41.98", "-87.9"],
              "Atlanta": ["33.66", "-84.42"],
              "Denver": ["39.87", "-104.67"]}
def weekly_forecast(lat, lon, option=None):
    weather_text.delete("1.0", "end")
    weather_text.insert(END, f"Weather for {option} \n")
    weather = requests.get(f"https://api.weather.gov/points/{lat},{lon}")

    json = weather.json()
    forecast = json["properties"]["forecast"]
    daily_forecast = requests.get(forecast).json()
    #CHANGE: Created a empty list to append the temperatures of the week
    temp_list=[]
    for section in daily_forecast["properties"]["periods"]:
        day = section["name"]
        temp = section["temperature"]
        #CHANGE: Adds the temperature to the temp_list list after changing it to int. 
        temp_list.append(int(temp))
        detail = section["detailedForecast"]
        text = f"{day} ---- {temp} \n"
        weather_text.insert(END, text)
        weather_text.tag_config("tag_name", justify="center")
        weather_text.tag_add("tag_name", "1.0", "end")
    #CHANGE: Calculates the average of the list. 
    temp_mean=sum(temp_list)/len(temp_list)
    #CHANGE: If else statements that determines what image, advice, and color to use
    if temp_mean <=60:
        img_path="C:/Users/zhang/Documents/GitHub/Lab-Excercises/Lab 7/coolgarfield.jpg"
        advice_text="The temperature this week is cold. You should stay in bed."
        color="blue"
    elif temp_mean <=80:
        img_path="C:/Users/zhang/Documents/GitHub/Lab-Excercises/Lab 7/normalgarfield.jpg"
        advice_text="The temperature this week is fair. You should still stay in bed though. "
        color="brown"
    else:
        img_path="C:/Users/zhang/Documents/GitHub/Lab-Excercises/Lab 7/hotgarfield.jpg"
        advice_text="The temperature this week is warm. Beach time!"
        color="red"
    #CHANGE: tells python that I'm refering to a globally defined variable which allows me to grid_forget() the 
    #previous labels. 
    #Reference: https://stackoverflow.com/questions/16424091/why-does-tkinter-image-not-show-up-if-created-in-a-function
    global garfield_img_label
    global garfield_text
    global label_garfield_advice
    #CHANGE: Try and except is used to run this code the first time because these labels haven't been created yet.
    try:
        garfield_img_label.grid_forget()
        garfield_text.grid_forget()
        label_garfield_advice.grid_forget()
    except:
        pass
    #CHANGE: Creates the label with text "Garfield's advice for this week"
    label_garfield_advice = Label(root, text = "Garfield's advice for this week: ")
    label_garfield_advice.grid()
    global garfield_image
    #CHANGE: opens the image and resizes it to 200,200 
    #Reference: https://stackoverflow.com/questions/6582387/image-resize-under-photoimage
    garfield_image=ImageTk.PhotoImage(Image.open(img_path).resize((200, 200)))
    garfield_img_label=Label(root,image=garfield_image)
    garfield_img_label.grid()
    #CHANGE: Creates the advice label with the advice_text and color
    garfield_text=Label(root, text=advice_text, fg=color)
    garfield_text.grid()
def chosen_option():
    option = city_select.get()
    value = city_latlon[option]
    lat = value[0]
    lon = value[1]
    weekly_forecast(lat,lon, option)

entry_frame = Frame(root)
entry_frame.grid(row=0, column=0)
drop_frame = Frame(root)
drop_frame.grid(row=0, column=1)

label_lat = Label(entry_frame , text="Enter Latitude").pack()
entry_lat = Entry(entry_frame , width=35, borderwidth=5, textvariable=lat).pack()
label_lon = Label(entry_frame , text="Enter Longitude").pack()
entry_lon = Entry(entry_frame , width=35, borderwidth=5, textvariable=lon).pack()
button = Button(entry_frame , font = 24, text = "Get Forecast", 
                command=lambda: weekly_forecast(lat.get(), lon.get()))
button.pack(pady = 20)

drop_label = Label(drop_frame, text="You can also select a city:").pack()
dropdown = OptionMenu(drop_frame, city_select, *city_list).pack()
choose_button = Button(drop_frame, text="Select City", command=chosen_option)
choose_button.pack(pady = 20)

weather_text = Text(root, height=16)
weather_text.grid(columnspan=2)

button2 = Button(root, font = 24, text = "Close Window", 
                command=root.destroy)
button2.grid(columnspan=2, pady = 20)

root.mainloop()

```
