```python
#imports the requests module
import requests
#imports the json module
import json
#from the tkinter library, import everything
from tkinter import *

#establish the window of the GUI and assigns it the variable, root
root = Tk()
#sets the size of the interface to 800 by 900
root.geometry("800x900") 
#Titles the window Weather App
root.title("Weather App")


#creates a string variable named lat 
lat = StringVar() # lat = "42.098701"
#Sets the string variable named lat to none
lat.set("")
#creates a string variable named lon
lon = StringVar() # lon = "-75.912537"
#Sets the string variable named lon to none
lon.set("")
#creates a string variable named city_select
city_select = StringVar()
#Sets the string variable named city_select to Select a City
city_select.set("Select a City")

#creates the list of cities, city_list, that will populate the dropdown menu. This list contains Binghamton, New York
#Chicago, Atlanta, and Denver. 
#source https://forecast.weather.gov/stations.php?foo=0
city_list = ["Binghamton", "New York", "Chicago", "Atlanta", "Denver"]

#creates a dictionary with the keys being the cities in the dropdown menu and the values being a list
#containing their respective lat and lon. 
#list syntax: [lat, lon]
city_latlon = {"Binghamton": ["42.09", "-75.91"],
               "New York": ["40.78", "-73.96"],
              "Chicago": ["41.98", "-87.9"],
              "Atlanta": ["33.66", "-84.42"],
              "Denver": ["39.87", "-104.67"]}

#creates a function named weekly_forecast with the optional parameters lat, lon
def weekly_forecast(lat, lon, option=None):
    
    #deletes any text in the weather_text text widget
    weather_text.delete("1.0", "end")
    #populates the weather_text text widget with the weather for whatever value option has. 
    weather_text.insert(END, f"Weather for {option} \n")
    #calls the api with the lat and lon values and assigns the call to variable weather.
    weather = requests.get(f"https://api.weather.gov/points/{lat},{lon}")

    #Creates a json file from the weather api call and assigns it to the variable json
    json = weather.json()
    #locates the json's key, properties, and its respective value and then from there, locates the key, forecast,
    #and its respective value and assigns it to the variable forecast
    forecast = json["properties"]["forecast"]
    #using the forecast variable, creates a new api call, makes a json file and assigns that json file to the variable
    #daily_forecast
    daily_forecast = requests.get(forecast).json()

    #creates a for loop that loops through all the variables in the value found under the key properties
    #and the key periods daily_forecast json file
    for section in daily_forecast["properties"]["periods"]:
        #Assigns the value associated with the key name to the variable day
        day = section["name"]
        #Assigns the value associated with the key temperature to the variable temp
        temp = section["temperature"]
        #Assigns the value associated with the key detailedForecast to the variable detail
        detail = section["detailedForecast"]
        #creates the text that is to be displayed with the day variable and temp variable and assigns it to variable text
        text = f"{day} ---- {temp} \n"
        #inserts the text variable to end of the existing text in the weather_text text widget. 
        weather_text.insert(END, text)
        #creates the tag tag_name and configures it so that it aligns the text to be centered
        weather_text.tag_config("tag_name", justify="center")
        #adds the tag_name tag for all text in the text widget weather_text
        weather_text.tag_add("tag_name", "1.0", "end")

#Creates a function chosen_option
def chosen_option():
    #gets the value in the variable city_select and assigns it to the variable option
    option = city_select.get()
    #using the variable option as a key, it grabs the lat and lon of the chosen city and assigns it to the variable
    #value. 
    value = city_latlon[option]
    #the first variable in the variable is assigned to the variable lat
    lat = value[0]
    #the second variable in the variable is assigned to the variable lon
    lon = value[1]
    #runs the weekly_forecast function with the variables lat and lon and option
    weekly_forecast(lat,lon, option)

#creates a frame in the GUI window root named entry_frame
entry_frame = Frame(root)
#places the frame in a grid format at coordinates 0,0
entry_frame.grid(row=0, column=0)
#creates a frame in the GUI window root named drop_frame
drop_frame = Frame(root)
#places the frame in a grid format at coordinates 0,1
drop_frame.grid(row=0, column=1)

#creates a label in the entry_frame with the text "Enter Latitude" and packs it in the frame and assigns it label_lat
label_lat = Label(entry_frame , text="Enter Latitude").pack()
#creates a entry in the entry_frame with a width of 35, borderwidth of 5 and assigns any entry to the variable lat and 
#packs it into the entry_frame and assigns all of this to entry_lat
entry_lat = Entry(entry_frame , width=35, borderwidth=5, textvariable=lat).pack()
#creates a label in the entry_frame with the text "Enter Longitude" and packs it in the frame and assigns it label_lon
label_lon = Label(entry_frame , text="Enter Longitude").pack()
#creates a entry in the entry_frame with a width of 35, borderwidth of 5 and assigns any entry to the variable lon and 
#packs it into the entry_frame and assigns all of this to entry_lon
entry_lon = Entry(entry_frame , width=35, borderwidth=5, textvariable=lon).pack()
#creates a button in the entry_frame with 24 size font and text "Get Forecast". It assigns command to the function
#weekly forecast with the values of lat and lon with the use of lambda and assigns all of this to button
button = Button(entry_frame , font = 24, text = "Get Forecast", 
                command=lambda: weekly_forecast(lat.get(), lon.get()))
#packs the button with a 20 padding
button.pack(pady = 20)

#creates the label in the drop_frame with the text "You can also select a city:" and packs it into the drop_frame
#and assigns all of this to drop_label
drop_label = Label(drop_frame, text="You can also select a city:").pack()
#creates the dropdown menu in the drop_frame, displaying the value of city_select and containing the variables in the
#list city_list and packs it in the drop_frame
dropdown = OptionMenu(drop_frame, city_select, *city_list).pack()
#Creates a button in the drop_frame with the text "select city" and excutes the function chosen_option and assigns
#all of this to the variable choosen_button
choose_button = Button(drop_frame, text="Select City", command=chosen_option)
#packs the choose_button with 20 padding
choose_button.pack(pady = 20)

#creates the text widget weather_text with the height of 16
weather_text = Text(root, height=16)
#places it in the grid format and makes it span across two column units
weather_text.grid(columnspan=2)

#creates a button in the root window with size 24 font, the text "Close Window" and command to destroy
button2 = Button(root, font = 24, text = "Close Window", 
                command=root.destroy)
#Places this button in a grid format with 20 padding and makes it span 2 column units
button2.grid(columnspan=2, pady = 20)

#makes the window loop until the user closes it. 
root.mainloop()
```
