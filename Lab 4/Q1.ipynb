```python
import requests
import json
```

```python
def get_def(word):
    word_dict=requests.get(f"https://api.dictionaryapi.dev/api/v2/entries/en/{word}")
    word_dict_json=word_dict.json()
    for i in word_dict_json[0]["meanings"]:
        for j in i["definitions"]:
            print(j["definition"])
```

```python
get_def("worry")
```

```python
def get_pronunciation(word):
    word_dict=requests.get(f"https://api.dictionaryapi.dev/api/v2/entries/en/{word}")
    word_dict_json=word_dict.json()
    return word_dict_json[0]["phonetic"]#[0]["definitions"][0]["definition"]
```

```python
get_pronunciation("worry")
```

```python
def get_examples(word):
    word_dict=requests.get(f"https://api.dictionaryapi.dev/api/v2/entries/en/{word}")
    word_dict_json=word_dict.json()
    for i in word_dict_json[0]["meanings"]:
        for j in i["definitions"]:
            try:
                print(j["example"])
            except:
                pass
```

```python
get_examples("square")
```

```python
def get_synonyms(word):
    word_dict=requests.get(f"https://api.dictionaryapi.dev/api/v2/entries/en/{word}")
    word_dict_json=word_dict.json()
    for i in word_dict_json[0]["meanings"]:
        for j in i["synonyms"]:
            print(j)
```

```python
get_synonyms("sad")
```
