```python
# github.py
#imports requests module
import requests

# REPLACE the following variables with your Client ID and Client Secret
#Assigns variable ClIENT_Id to the string value 94b5bc4be3b182015103
CLIENT_ID = "94b5bc4be3b182015103"
#Assigns variable CLIENT_SECRET to the string value 6a0bb4cc534e6485d4050e6527b0ed435c0d739f
CLIENT_SECRET = "6a0bb4cc534e6485d4050e6527b0ed435c0d739f"

# REPLACE the following variable with what you added in the
# "Authorization callback URL" field
#This is the redirect url that the api will use to send information to me. This url simply outputs what it gets as an input.
REDIRECT_URI = "https://httpbin.org/anything"
```

```python
# github.py

# ...
#function that creates the oauth link
def create_oauth_link():
    #Creates a dictionary with parameters
    params = {
        #The key client id and its corresponding value 
        "client_id": CLIENT_ID,
        #The key redirect url and its corresponding value        
        "redirect_uri": REDIRECT_URI,
        #The key scope and it is defined as user
        "scope": "user",
        #The key response type and it is defined as code
        "response_type": "code",
    }
    #assigns the api call to variable endpoint
    endpoint = "https://github.com/login/oauth/authorize"
    #Calls the api with the parameters after we set 
    response = requests.get(endpoint, params=params)
    #return the api call
    return response.url
```

```python
# github.py

# ...
#creates a function that returns access token when the user inputs code
def exchange_code_for_access_token(code=None):
    #defines parameters
    params = {
        #The key client id and its corresponding value 
        "client_id": CLIENT_ID,
        #The key client_secret and its corresponding value 
        "client_secret": CLIENT_SECRET,
        #The key redirect url and its corresponding value 
        "redirect_uri": REDIRECT_URI,
        #The key code and its corresponding value
        "code": code,
    }
    #creates dictionary with accept key
    headers = {"Accept": "application/json"}
    #api call named endpoint
    endpoint = "https://github.com/login/oauth/access_token"
    #actually calls the api and assign it to value reponse
    response = requests.post(endpoint, params=params, headers=headers).json()
    #returns the access token from the api call
    return response["access_token"]
```

```python
# github.py

# ...
#creates the oauth_link and assigns it to variable link
link = create_oauth_link()
#prints out text for users to follow
print(f"Follow the link to start the authentication with GitHub: {link}")
#gets input from user and assigns it to code
code = input("GitHub code: ")
#gets access token using exchange code for access token function and the variable code
access_token = exchange_code_for_access_token(code)
#Prints out access token and the code used to get it
print(f"Exchanged code {code} with access token: {access_token}")
```

```python
# github.py

# ...
#Creates function that prints the user's info with the optional parameter access token
def print_user_info(access_token=None):
    #dictionary with key authorization and defined as the access token as a string value
    headers = {"Authorization": f"token {access_token}"}
    #api call named endpoint
    endpoint = "https://api.github.com/user"
    #calls the api and makes a json file
    response = requests.get(endpoint, headers=headers).json()
    #finds the name and assigns it to a name variable
    name = response["name"]
    #Finds the username and assignsit to a username variable
    username = response["login"]
    #Finds the number of private repos and assigns it to a variable
    private_repos_count = response["total_private_repos"]
    #Prints out all the variables
    print(
        f"{name} ({username}) | private repositories: {private_repos_count}"
    )

# ...
```
